Load the file with pickle

```python
with open('Ionic_salts_data.pickle', 'rb') as inp:
    cat_dict, an_dict = pickle.load(inp)
```

**Select  cation and anion**

```python
cation_selection = widgets.Dropdown(options = cat_dict.keys(), value = 'Na')
cation_selection
```

```python
anion_selection = widgets.Dropdown(options = an_dict.keys(), value = 'Cl')
anion_selection
```

```python
cation = cation_selection.value
anion = anion_selection.value
```

```python
MW = cat_dict[cation]['MW, g/mol']*an_dict[anion]['n'] + an_dict[anion]['MW, g/mol']*cat_dict[cation]['p']
cat_aff = cat_dict[cation]['Rel.Affin'] 
an_aff = an_dict[anion]['Rel.Affin'] 
```

Selection of surfactant type and calculation of affinity and equivalent salnity

```python
surfactant_type_selection = widgets.Dropdown(options = ['', 'Cationic', 'Anionic'], value = '')
surfactant_type_selection
```

```python
surfactant_type = surfactant_type_selection.value
```

```python
if surfactant_type == 'Cationic':
    aff = (1/(an_dict[anion]['n']*cat_dict[cation]['p']))/an_aff
elif surfactant_type == 'Anionic':
    aff = aff = (1/(an_dict[anion]['n']*cat_dict[cation]['p']))/cat_aff
    

Salinity = float(input('Enter your salinity in g/100 mL'))
M = Salinity/MW
S_eq = M/aff*58.5

print('Equivalent salinity in terms of g NaCl/100 mL is ', S_eq)
```
