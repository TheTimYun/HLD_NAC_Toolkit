Load the data files

```python
with open('Nonionic_salts_data.pickle', 'rb') as inp:
    cat_dict, an_dict = pickle.load(inp)
```

**Select cation and anion**

```python
cation_selection = widgets.Dropdown(options = cat_dict.keys(), value = 'Na')
cation_selection
```

```python
anion_selection = widgets.Dropdown(options = an_dict.keys(), value = 'Cl')
anion_selection
```

Calculation of affinity and equivalent salinity

```python
cation = cation_selection.value
anion = anion_selection.value
```

```python
MW = cat_dict[cation]['MW, g/mol']*an_dict[anion]['n'] + an_dict[anion]['MW, g/mol']*cat_dict[cation]['p']
cat_bf = cat_dict[cation]['B'] 
an_bf = an_dict[anion]['B'] 
```

```python
BMod = (cat_bf * cat_dict[cation]['p'] + an_bf * an_dict[anion]['n'])/cat_dict[cation]['p']
```

```python
Salinity = float(input('Enter your salinity in g/100 mL'))
b = 96*BMod/MW
S_eq = (Salinity*b)/0.13
print(S_eq)
```
