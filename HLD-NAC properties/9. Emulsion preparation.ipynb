If you want to perform emulsion preparation with catastrophic inversion, you can use this notebook, which will show you the optimal range of properties for the preparation

```python
import numpy as np
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
#import matplotlib.pyplot as plt
#import matplotlib.patches as patches
from functions_for_surfactants import  HLD_calculation, draw_emulsion_plot
```

```python
Cc = float(input('Enter Cc value'))
Sal = float(input('Enter salinity value'))
Temp = float(input('Enter temperature value'))
EACN = float(input('Enter EACN of your oil'))
```

```python
surfactant_selection = widgets.Dropdown(options = ['', 'Ionic', 'Nonionic', 'Zwitterionic'], value = '')
surfactant_selection
```

```python
if surfactant_selection.value == 'Ionic':
    conc = float(input('Enter connetration of surfactant in mol/L'))
    MW = float(input('Enter MW of the counterion'))
    Cor = 0.3 * (conc/10)*MW
else:
    Cor = 0

HLD = HLD_calculation(EACN = EACN, Cc = Cc, Sal = Sal, Temp = Temp, Type = surfactant_selection.value, Cor = Cor)
print('HLD for given parameters is {}'.format(HLD))
```

Plot below can guide you through the emulsion preparation. Detailed explanation can be found in the [book](https://www.stevenabbott.co.uk/practical-surfactants/inversion.php) by Steven Abbott. Yellow paths designate water-in-oil emulsions, whereas blue paths designate oil-in-water emulsions, triangles are transition zones. Dashed green lines designate range for optimal HLD for emulsification, which is between 2 and 3 for both poisitive and negative sides. Your calculated HLD is drawn with red line. Therefore you can track where you are in the formulation space

```python
fig = draw_emulsion_plot(HLD_val = HLD)
```
