```python
from rdkit import Chem
import rdkit
import ipywidgets as widgets
import numpy as np
from functions_for_surfactants import len_calculation

p1 = '[CX4H3][CX4H2]'
p2 = '[CX4H3][CX4H1]([#6])[#6]'
secondary = '[CX4H2;!$(C(-O)-C-O)]'
tertiary_2 = '[CH3][CH1;X4]([CH3])'
tertiary = '[CH1;X4]'
benzene = 'c1ccccc1'
EO = 'O[CH2][CH2]O'
PO = 'OC(C)CO'
```

```python
nonionic_type = widgets.Dropdown(options = ['Ethoxylated alcohol', 'Ethoxylated phenol'])
nonionic_type
```

```python
input_type = widgets.Dropdown(options = ['SMILES', 'Manual'])
input_type
```

```python
if input_type.value == 'Manual':
    n_EOs = int(input('Enter number of EOs units'))
    n_C = int(input('Enter number of carbon atoms'))       
elif input_type.value == 'SMILES':
    smi = input('Enter SMILES of you molecule')
    mol = Chem.MolFromSmiles(smi)
    L = len_calculation(mol, include_benzene=False)
    n_C = 1 + ((L/1.3)-1.5)/1.265

if nonionic_type.value == 'Ethoxylated alcohol':
     Cc = 0.2 * n_C - n_EOs + 2.7

elif nonionic_type.value == 'Ethoxylated phenol':
    HLB =  20 * ( (44.05 * n_EOs)/(44.05*n_EOs + 14 * n_C + 94))
    Cc  = (11.5 - HLB)/0.85
```

```python
print('Cc of the surfactant is: ', round(Cc,3))
cloud_point = 25 + (-Cc + (0.16 * 7.3))/0.06

print('Cloud point for the surfactant is (in degC)L ', round(cloud_point,2))
```
