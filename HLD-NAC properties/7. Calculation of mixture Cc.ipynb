```python
from rdkit.Chem.Descriptors import MolWt
from rdkit import Chem
import ipywidgets as widgets
import numpy as np
```

```python
concentration_type = widgets.Select(options = ['Molar', 'Weight'])
concentration_type
```

```python
number_of_surfactants = int(input('Enter number of surfactants species'))
```

```python
CCs = []
for i in range(number_of_surfactants):
    CCs.append(float(input('Enter Cc value of surfactant {}'.format(i+1))))
```

```python
ratios = []
for i in range(number_of_surfactants):
    ratios.append(float(input('Enter ratio of surfactant {}, {}'.format(i, concentration_type.value))))
```

If ypu know SMILES, use the code below. If you know molecular weight, use underlying cell

```python
if concentration_type.value == 'Molar':
    average_cc = np.sum(np.array(CCs)*np.array(ratios))
else:
    smis = []
    for i in range(number_of_surfactants):
        smis.append(input('Enter SMILES surfactant {}'.format(i+1)))
    MolWts = [MolWt(Chem.MolFromSmiles(smi)) for smi in smis]
    mols = 100*np.array(ratios)/MolWts
    molar_ratios = mols/np.sum(mols)
    average_cc = np.sum(np.array(CCs)*np.array(molar_ratios))
print('Cc of the studied system is {}'.format(average_cc))

```

Use the cell below if you know molecular weights of your surfactants

```python
if concentration_type.value == 'Molar':
    average_cc = np.sum(np.array(CCs)*np.array(ratios))
else:
    MolWts = []
    for i in range(number_of_surfactants):
        MolWts.append(float(input('Enter MW of surfactant {}'.format(i+1))))
    mols = 100*np.array(ratios)/MolWts
    molar_ratios = mols/np.sum(mols)
    average_cc = np.sum(np.array(CCs)*np.array(molar_ratios))
print('Cc of the studied system is {}'.format(average_cc))
```
