**Overview**  
This notebook describes folloiwing stages:
1. Data import and reading;
2. Data dividing into ionics, non-ionics and extended;
4. Calculation of descriptors;
5. Dropping the most correlated features;
6. Export of features and targets for further ML models training

```python
from rdkit import RDConfig
from tqdm import tqdm
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem
from copy import copy
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os 
from sklearn.decomposition import PCA
import datamol as dm
import molfeat

from padelpy import from_smiles
from molfeat.calc.descriptors import RDKitDescriptors2D, RDKitDescriptors3D
from molfeat.trans import MoleculeTransformer
import mols2grid
from molfeat_padel.calc import PadelDescriptors
tqdm.pandas()

import pickle
from sklearn.impute import KNNImputer
from feature_engine.selection import DropCorrelatedFeatures 
```

First, let's read the dataset. It's organized in Excel table. "Name" columnis the surfactant name, "SMILES" column is SMILES, "Cc" is critical curvature value, "Upper" and "Lower" boundaries - are uncertainty limits (if stated in reference), "Is ionic" and "Is non-ionic" points to the chemical nature, "Is sure" is used to denote surfactants, Cc of which was recalculated or if chemical formula is unclear or something else. Fortunately, there are not so many instance of them. 



First, let's read the data separately for ionic, non-ionics and extended (that are denoted with ones in both columns)

```python
Data = pd.read_excel('Dataset_without_reference.xlsx')
Data.head()
```

```python
Data_anionic = Data.query('Is_anionic==1')
Data_cationic = Data.query('Is_cationic==1')
Data_non_ionic = Data.query('Is_non_ionic==1')
Data_extended = Data.query('Is_extended==1')
```

Thre can be a multiple instances of the same compound. To prevent from using them several times, we canonicalize SMILES and then groupby table.

```python
def canonicalize_SMILES(smi):
    can_smi = Chem.MolToSmiles(Chem.MolFromSmiles(smi))
    return can_smi
```

```python
Data_anionic.loc[:, 'SMILES'] = Data_anionic['SMILES'].progress_apply(canonicalize_SMILES)
Data_cationic.loc[:, 'SMILES'] = Data_cationic['SMILES'].progress_apply(canonicalize_SMILES)
Data_non_ionic.loc[:, 'SMILES'] = Data_non_ionic['SMILES'].progress_apply(canonicalize_SMILES)
Data_extended.loc[:, 'SMILES'] = Data_extended['SMILES'].progress_apply(canonicalize_SMILES)
```

```python
def custom_agg(series):
    if series.notna().sum() > 1:
        return series.dropna().mean()
    elif series.notna().sum() == 1:
        return series.dropna().iloc[0]
    else:
        return np.nan
```

```python
Data_anionic = Data_anionic.groupby(by = 'SMILES').agg({'Cc ':'mean', 'Upper':custom_agg, 'Lower':custom_agg}).reset_index(drop = False)
Data_cationic = Data_cationic.groupby(by = 'SMILES').agg({'Cc ':'mean', 'Upper':custom_agg, 'Lower':custom_agg}).reset_index(drop = False)
Data_non_ionic = Data_non_ionic.groupby(by = 'SMILES').agg({'Cc ':'mean', 'Upper':custom_agg, 'Lower':custom_agg}).reset_index(drop = False)
Data_extended = Data_extended.groupby(by = 'SMILES').agg({'Cc ':'mean', 'Upper':custom_agg, 'Lower':custom_agg}).reset_index(drop = False)
```

```python
print('Number of instances for anionic surfactants is {}'.format(len(Data_anionic)))
print('Number of instances for cationic surfactants is {}'.format(len(Data_cationic)))
print('Number of instances for non-ionic surfactants is {}'.format(len(Data_non_ionic)))
print('Number of instances for extended surfactants  is {}'.format(len(Data_extended)))
```

Then we save all instnces into several csv files

```python
Data_anionic.to_csv('Data_anionic.csv')
Data_cationic.to_csv('Data_cationic.csv')
Data_non_ionic.to_csv('Data_non_ionic.csv')
Data_extended.to_csv('Data_extended.csv')
```

```python
Data_anionic = pd.read_csv('Data_anionic.csv', index_col = 0)
Data_cationic = pd.read_csv('Data_cationic.csv', index_col = 0)
Data_non_ionic = pd.read_csv('Data_non_ionic.csv', index_col = 0)
Data_extended = pd.read_csv('Data_extended.csv', index_col=0)
```

Then we get *mol* column for every surfactant in their corresponding data files

**Descriptors calculation**

We create a function to get conformers of molecule

```python
def embed_optimize(smi):
    try:
        m = Chem.MolFromSmiles(smi)
        m = Chem.AddHs(m)
        params = AllChem.ETKDGv3()
        params.maxIterations = 1000
        params.useRandomCoords = True
        AllChem.EmbedMolecule(m, params)
        AllChem.MMFFOptimizeMolecule(m, maxIters=500)
        return m
    except:
        print(smi)
```

And then get *mol* column

```python
Data_anionic['mol'] = Data_anionic.SMILES.progress_apply(embed_optimize)
Data_cationic['mol'] = Data_cationic.SMILES.progress_apply(embed_optimize)
Data_non_ionic['mol'] = Data_non_ionic.SMILES.progress_apply(embed_optimize)
Data_extended['mol'] = Data_extended.SMILES.progress_apply(embed_optimize)
```

Then we create instances of descriptors calculators and molecule transformers and calculate 2D and 3D RDKit descriptors

```python
calc_2d = RDKitDescriptors2D(ignore_descrs=['AvgIpc', 'Ipc'], do_not_standardize=True)
calc_3d = RDKitDescriptors3D()
trans_2D = MoleculeTransformer(calc_2d, verbose = True)
trans_3D = MoleculeTransformer(calc_3d, verbose = True)
```

And then we calculate 2D and 3D descriptors as concatenate them into one DataFrame

```python
rdkit_descs = {}
with dm.without_rdkit_log():
    for data, name in zip([Data_anionic, Data_cationic, Data_non_ionic, Data_extended], ['RDKIt_anionic', 'RDKit_cationic', 'RDKIt_non_ionic', 'RDKIt_extended']):
        print(name)
        feats_2D = trans_2D(data.SMILES.values)
        feats_3D = trans_3D(data.mol.values)
        rdkit_descs[name] = pd.DataFrame(np.concatenate([feats_2D, feats_3D], axis  = 1), columns=calc_2d.columns + calc_3d.columns).drop(columns='Alerts')
        
```

Let's check for nulls

```python
print('Nulls for anionic descriptors', rdkit_descs['RDKIt_anionic'].isnull().sum().sum())
print('Nulls for cationic descriptors', rdkit_descs['RDKit_cationic'].isnull().sum().sum())
print('Nulls for non-ionic descriptors', rdkit_descs['RDKIt_non_ionic'].isnull().sum().sum())
print('Nulls for extended descriptors', rdkit_descs['RDKIt_extended'].isnull().sum().sum())
```

For non-ionic and extended surfactant we also add two columns - number of EOs and POs

```python
smarts_PO = 'OC(C)CO'
smarts_EO = 'O[CH2][CH2]O'
PO_frag = Chem.MolFromSmarts(smarts_PO)
EO_frag = Chem.MolFromSmarts(smarts_EO)
```

```python

for data, name in zip([Data_non_ionic, Data_extended],['RDKIt_non_ionic', 'RDKIt_extended']):
    num_EOs, num_POs = [], []
    for mol in data.mol:
        num_EO = len(mol.GetSubstructMatches(EO_frag))
        num_PO = len(mol.GetSubstructMatches(PO_frag))
        num_EOs.append(num_EO)
        num_POs.append(num_PO)
    rdkit_descs[name]['EOs'] = np.array(num_EOs)
    rdkit_descs[name]['POs'] = np.array(num_POs)
```

Finally, for every set of descriptors we remove values of high correlation and save names of features, that we dropped

```python
features_to_drop_rdkit = {}

for k, v in rdkit_descs.items():
    dropper = DropCorrelatedFeatures(threshold=0.8)
    print('Initial number of descriptors for', k, 'is ', len(v.columns))
    #some of columns after descriptors calculation are duplicated
    v = v.loc[:, ~v.columns.duplicated()].copy()
    v_dropped = dropper.fit_transform(v)
    rdkit_descs[k] = v_dropped
    print('Final number of descriptors for', k, 'is ', len(v_dropped.columns))
    features_to_drop_rdkit[k] = dropper.features_to_drop_
```

Then we dump everything to pickle file

```python
with open('rdkit_descs.pickle', 'wb') as out:
    pickle.dump(rdkit_descs, out)
```

```python
with open('features_to_drop_rdkit.pickle', 'wb') as out:
    pickle.dump(features_to_drop_rdkit, out)
```

Our next step is the calculation of **[PaDEL descriptors](https://onlinelibrary.wiley.com/doi/full/10.1002/jcc.21707)** descriptors, which are very useful for prediction of molecular properties. To do it, we use add-on for molfeat.

```python
from collections import defaultdict
```

```python
padel_descs = {}
for data, name in zip([Data_anionic, Data_cationic, Data_non_ionic, Data_extended], ['PaDEL_anionic', 'PaDEL_cationic', 'PaDEL_non_ionic', 'PaDEL_extended']):
    print(name)
    d = defaultdict(list)
    for smi in tqdm(data.SMILES):
        tmp = from_smiles(smi)
        for k, v in tmp.items():
            d[k].append(v)
    padel_descs[name] = pd.DataFrame(d)
    for column in padel_descs[name].columns:
        padel_descs[name][column] = pd.to_numeric(padel_descs[name][column], errors= 'coerce')
```

However, there are a lot of nulls

```python
print('Nulls for anionic descriptors', padel_descs['PaDEL_anionic'].isnull().sum().sum())
print('Nulls for cationic descriptors', padel_descs['PaDEL_cationic'].isnull().sum().sum())
print('Nulls for non-ionic descriptors', padel_descs['PaDEL_non_ionic'].isnull().sum().sum())
print('Nulls for extended descriptors', padel_descs['PaDEL_extended'].isnull().sum().sum())
```

So we use simple KNN imouter to fill in th gaps

```python
imputers_for_padel = {}

for k, v in padel_descs.items():
    imputer = KNNImputer()
    desc_names = v.columns
    padel_descs[k] = pd.DataFrame(imputer.fit_transform(v), columns = desc_names)
    imputers_for_padel[k] = imputer
```

And dump all imputers in pickle file

```python
with open('imputers_for_padel.pickle', 'wb') as out:
    pickle.dump(imputers_for_padel, out)
```

Again we add columns of EOs and POs

```python
for data, name in zip([Data_non_ionic, Data_extended],['PaDEL_non_ionic', 'PaDEL_extended']):
    num_EOs, num_POs = [], []
    for mol in data.mol:
        num_EO = len(mol.GetSubstructMatches(EO_frag))
        num_PO = len(mol.GetSubstructMatches(PO_frag))
        num_EOs.append(num_EO)
        num_POs.append(num_PO)
    padel_descs[name]['EOs'] = np.array(num_EOs)
    padel_descs[name]['POs'] = np.array(num_POs)
```

And drop features of high correlation

```python
features_to_drop_padel = {}

for k, v in padel_descs.items():
    dropper = DropCorrelatedFeatures(threshold=0.8)
    print('Initial number of descriptors for', k, 'is ', len(v.columns))
    v_dropped = dropper.fit_transform(v)
    padel_descs[k] = v_dropped
    print('Final number of descriptors for', k, 'is ', len(v_dropped.columns))
    features_to_drop_padel[k] = dropper.features_to_drop_
```

Then we save PaDEL descriptors and names of PaDEL features, that we dropped

```python
with open('padel_descks.pickle', 'wb') as out:
    pickle.dump(padel_descs, out)
```

```python
with open('featured_to_drop_padel.pickle', 'wb') as out:
    pickle.dump(features_to_drop_padel, out)
```
