```python
import pandas as pd
import numpy as np
import pickle
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import GridSearchCV, train_test_split, cross_val_score, cross_val_predict
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score, root_mean_squared_error, mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.svm import SVR
from hyperopt import tpe, Trials, hp, fmin, STATUS_OK
from functools import partial
from sklearn.neighbors import KNeighborsRegressor
```

```python
with open ('Features_PaDEL/padel_descks.pickle', 'rb') as inp:
    rdkit_descs = pickle.load(inp)
```

```python
nonionic_descs = rdkit_descs['PaDEL_extended']
```

```python
target = pd.read_csv('Data/Data_extended.csv')['Cc ']
```

```python
X_train, X_test, y_train, y_test = train_test_split(nonionic_descs, target, random_state=0)
```

Images for extended

```python
pipe_svm = make_pipeline(StandardScaler(), SVR(C= 0.03816646541480999, coef0 = 0.670962401453654, degree = 4, gamma ='scale', kernel = 'linear'))
```

```python
pipe_svm.fit(X_train, y_train)
```

```python
train_predicted = pipe_svm.predict(X_train)
test_predicted = pipe_svm.predict(X_test)
```

```python
plt.figure(figsize = (8, 8))
plt.plot(y_train, train_predicted, 'ro', label = 'Train values')
plt.plot(y_test, test_predicted, 'go', label = 'Test values')
plt.plot([target.min()-5, target.max()+5], [target.min()-5, target.max()+5])
plt.xlim([-5, 2])
plt.ylim([-5, 2])
plt.legend()
plt.xlabel('True values')
plt.ylabel('Predicted values')
plt.savefig('SVM-extended.png')
```
