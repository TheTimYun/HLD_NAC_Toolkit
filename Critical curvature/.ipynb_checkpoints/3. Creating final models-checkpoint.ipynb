```python
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.linear_model import Ridge
from sklearn.metrics import root_mean_squared_error, r2_score, mean_absolute_error
import molfeat
from molfeat.calc.descriptors import RDKitDescriptors2D, RDKitDescriptors3D
from molfeat.trans import MoleculeTransformer
from molfeat_padel.calc import PadelDescriptors
from padelpy import from_smiles
from sklearn.impute import KNNImputer
from functions import embed_optimize

import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt

from rdkit import Chem
from rdkit.Chem import AllChem


with open('Features_RDKit/features_to_drop_rdkit.pickle', 'rb') as inp:
    features_to_drop_rdkit = pickle.load(inp)

with open('Features_PaDEL/featured_to_drop_padel.pickle', 'rb') as inp:
    features_to_drop_padel = pickle.load(inp)

with open('Features_United/features_to_drop_united.pickle', 'rb') as inp:
    features_to_drop_united = pickle.load(inp)


with open('Features_PaDEL/imputers_for_padel.pickle', 'rb') as inp:
    imputers = pickle.load(inp)
```

**Non-ionic surfactant**

First, let's load the dataset with descriptors

```python
with open('rdkit_descs.pickle', 'rb') as inp:
    descs = pickle.load(inp)

target = pd.read_csv('Data_non_ionic.csv')['Cc ']
```

```python
descs_non_ionic = descs['RDKIt_non_ionic']
```

```python
pipe = make_pipeline(StandardScaler(), SVR(C = 0.015082446203887703, coef0 = 0.5909033059574486, degree = 2, gamma = 'auto', kernel = 'linear'))
```

```python
loo = LeaveOneOut()
predictions = cross_val_predict(pipe, X = descs_non_ionic, y = target, cv = loo)
print('RMSE for loo is {} '.format(root_mean_squared_error(target, predictions)))
print('MAE for loo is {} '.format(mean_absolute_error(target, predictions)))
print('R2 for loo is {} '.format(r2_score(target, predictions)))
```

```python
pipe.fit(descs_non_ionic, target)
```

```python
pipe_lr = make_pipeline(StandardScaler(), Ridge(alpha = 98.77341099629008, solver = 'lsqr'))
```

```python
loo = LeaveOneOut()
predictions = cross_val_predict(pipe_lr, X = descs_non_ionic, y = target, cv = loo)
print('RMSE for loo - Ridge regression is {} '.format(root_mean_squared_error(target, predictions)))
print('MAE for loo - Ridge regression is {} '.format(mean_absolute_error(target, predictions)))
print('R2 for loo - Ridge regression is {} '.format(r2_score(target, predictions)))
```

```python
(target - predictions).abs().std(ddof = 0)
```

```python
pipe_lr.fit(descs_non_ionic, target)
```

```python
with open('models_for_prediction_non_ionic.pickle', 'wb') as out:
    pickle.dump([pipe, pipe_lr], out)
```

let's find relative importances for each feature

```python
fig = plt.figure(figsize = (12,8))
reg_lr = pipe_lr.steps[1][1]
plt.barh(width = reg_lr.coef_[np.argsort(np.abs(reg_lr.coef_))[-15:]], 
         y = descs_non_ionic.columns[np.argsort(np.abs(reg_lr.coef_))[-15:]])
plt.xlabel('Coefficient values')
```

In the next section, you can easily calculate Cc of given surfactant

```python
smi = input('Enter you molecule SMILES   ')
mol = Chem.MolFromSmiles(smi)
optimized_molecule = embed_optimize(smi)
```

```python
calc_2d = RDKitDescriptors2D(ignore_descrs=['AvgIpc', 'Ipc'], do_not_standardize=True)
calc_3d = RDKitDescriptors3D()
trans_2D = MoleculeTransformer(calc_2d, verbose = True)
trans_3D = MoleculeTransformer(calc_3d, verbose = True)

feats_2d = trans_2D(smi)
feats_3d = trans_3D(optimized_molecule)

feats = pd.DataFrame(np.concatenate([feats_2d, feats_3d], axis = 1), columns = calc_2d.columns + calc_3d.columns).drop(columns = 'Alerts')
smarts_PO = 'OC(C)CO'
smarts_EO = 'O[CH2][CH2]O'
PO_frag = Chem.MolFromSmarts(smarts_PO)
EO_frag = Chem.MolFromSmarts(smarts_EO)
feats['EOs'] = len(mol.GetSubstructMatches(EO_frag))
feats['POs'] = len(mol.GetSubstructMatches(PO_frag))
feats.drop(columns = features_to_drop_rdkit['RDKIt_non_ionic'], inplace=True)

print('Prediction for non-ionic according to SVR is {}'.format(pipe.predict(feats)[0]))
print('Prediction for non-ionic according to LR is {}'.format(pipe_lr.predict(feats)[0]))
```

**After that, we will do everything for anionics**

```python
with open('padel_descks.pickle', 'rb') as inp:
    descs_anionic = pickle.load(inp)['PaDEL_anionic']
```

```python
target = pd.read_csv('Data_anionic.csv')['Cc ']
```

Ridge Regression

```python
pipe_lr = make_pipeline(StandardScaler(), Ridge(solver = 'saga', alpha = 69.95332303268256))
loo = LeaveOneOut()
predictions = cross_val_predict(pipe_lr, X = descs_anionic, y = target, cv = loo)
print('RMSE for loo is {} '.format(root_mean_squared_error(target, predictions)))
print('MAE for loo is {} '.format(mean_absolute_error(target, predictions)))
print('R2 for loo is {} '.format(r2_score(target, predictions)))
```

```python
(target - predictions).abs().std(ddof = 0)
```

```python
pipe_lr.fit(descs_anionic, target)
```

```python
fig = plt.figure(figsize = (12,8))
reg = pipe_lr.steps[1][1]
plt.barh(width = reg.coef_[np.argsort(np.abs(reg.coef_))[-15:]], 
         y = descs_anionic.columns[np.argsort(np.abs(reg.coef_))[-15:]])
plt.xlabel('Coefficient values')
```

SVR

```python
pipe = make_pipeline(StandardScaler(), SVR(C =  7.152758833526424, coef0 = 2.537414014288427, degree = 3, gamma = 'auto',  kernel = 'rbf'  ))
loo = LeaveOneOut()
predictions = cross_val_predict(pipe, X = descs_anionic, y = target, cv = loo)
print('RMSE for loo is {} '.format(root_mean_squared_error(target, predictions)))
print('MAE for loo is {} '.format(mean_absolute_error(target, predictions)))
print('R2 for loo is {} '.format(r2_score(target, predictions)))
```

```python
pipe.fit(descs_anionic, target)
```

```python
with open('models_for_prediction_anionic.pickle', 'wb') as out:
    pickle.dump([pipe, pipe_lr], out)
```

```python
smi = input('Enter you molecule SMILES   ')

feats = from_smiles(smi)
feats = pd.DataFrame(feats, index = [0])
for column in feats.columns:
    feats[column] = pd.to_numeric(feats[column], errors = 'coerce')

feats = pd.DataFrame(imputers['PaDEL_anionic'].transform(feats), columns = feats.columns)
feats.drop(columns = features_to_drop_padel['PaDEL_anionic'], inplace=True)
print('Prediction for anionic according to SVR model is {}'.format(pipe.predict(feats)[0]))
print('Prediction for anionic according to Ridge model is {}'.format(pipe_lr.predict(feats)[0]))
```

**Then we deal with cationic surfactants**

```python
with open('united_descs.pcikle', 'rb') as inp:
    descs_cationic = pickle.load(inp)['United_cationic']
```

```python
target = pd.read_csv('Data_cationic.csv')['Cc ']
```

```python
pipe_lr = make_pipeline(StandardScaler(), Ridge(solver = 'sparse_cg', alpha = 0.00010008528502641546))
loo = LeaveOneOut()
predictions = cross_val_predict(pipe_lr, X = descs_cationic, y = target, cv = loo)
print('RMSE for loo is {} '.format(root_mean_squared_error(target, predictions)))
print('MAE for loo is {} '.format(mean_absolute_error(target, predictions)))
print('R2 for loo is {} '.format(r2_score(target, predictions)))
```

```python
(target - predictions).abs().std(ddof = 0)
```

```python
pipe_lr.fit(descs_cationic, target)
```

```python
fig = plt.figure(figsize = (12,8))
reg = pipe_lr.steps[1][1]
plt.barh(width = reg.coef_[np.argsort(np.abs(reg.coef_))[-15:]], 
         y = descs_cationic.columns[np.argsort(np.abs(reg.coef_))[-15:]])
plt.xlabel('Coefficient values')
```

```python
with open('models_for_prediction_cationic.pickle', 'wb') as out:
    pickle.dump([pipe_lr], out)
```

```python
smi = input('Enter you molecule SMILES   ')

feats_padel = from_smiles(smi)
feats_padel = pd.DataFrame(feats_padel, index = [0])
for column in feats_padel.columns:
    feats_padel[column] = pd.to_numeric(feats_padel[column], errors = 'coerce')

feats_padel = pd.DataFrame(imputers['PaDEL_cationic'].transform(feats_padel), columns = feats_padel.columns)
feats_padel.drop(columns = features_to_drop_padel['PaDEL_cationic'], inplace=True)

optimized_molecule = embed_optimize(smi)
calc_2d = RDKitDescriptors2D(ignore_descrs=['AvgIpc', 'Ipc'], do_not_standardize=True)
calc_3d = RDKitDescriptors3D()
trans_2D = MoleculeTransformer(calc_2d, verbose = True)
trans_3D = MoleculeTransformer(calc_3d, verbose = True)
feats_2d = trans_2D(smi)
feats_3d = trans_3D(optimized_molecule)

feats_rdkit = pd.DataFrame(np.concatenate([feats_2d, feats_3d], axis = 1), columns = calc_2d.columns + calc_3d.columns).drop(columns = 'Alerts')
feats_rdkit.drop(columns = features_to_drop_rdkit['RDKit_cationic'], inplace=True)

feats = pd.concat([feats_padel, feats_rdkit], axis = 1)
feats.drop(columns = features_to_drop_united['United_cationic'], inplace=True)

print('Prediction for anionic according to Ridge model is {}'.format(pipe_lr.predict(feats)[0]))
```
