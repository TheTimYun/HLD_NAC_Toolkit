```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_anionic = pd.read_csv('Data_anionic.csv', usecols=[1,2,3,4]).assign(Type = 'Anionic')
data_cationic = pd.read_csv('Data_cationic.csv',usecols=[1,2,3,4]).reset_index(drop = True).assign(Type = 'Cationic')
data_non_ionic = pd.read_csv('Data_non_ionic.csv', usecols=[1,2,3,4]).assign(Type = 'Non_ionic')
data_extended = pd.read_csv('Data_extended.csv',usecols=[1,2,3,4]).reset_index(drop = True).assign(Type = 'Extended')
```

```python
cat = pd.concat([data_anionic, data_cationic, data_non_ionic, data_extended]).reset_index(drop = True)
```

```python
sns.boxplot(data = cat, x = 'Cc ', y = 'Type', hue = 'Type')
plt.savefig('Boxplot_cc.png')
```

```python
sns.violinplot(data= cat, x = 'Cc ', y = 'Type', hue = 'Type')
plt.xlim(-10,8)
plt.savefig('Violinplot_cc.png')
```
