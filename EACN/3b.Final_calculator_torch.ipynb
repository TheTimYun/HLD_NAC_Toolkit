**Overview**  
This notebook was created only to caclulate EACN on the base of SGIR model predictions. Just run all cells

```python
from torch_molecule import SGIRMolecularPredictor
```

```python
sgir = SGIRMolecularPredictor()
sgir.load_from_local('sgir_final.pt')
```

Here calculator starts!

```python
smi = input('Please, enter SMILES of your molecule')

EACN = sgir.predict([smi])['prediction']

print('EACN of the molecule is {}'.format(EACN[0]))
```
