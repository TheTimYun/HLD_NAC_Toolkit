**Overview**  
This notebook describes folloiwing stages:
1. Creating functions to:
    - perform search for optimal algorithm parameters for both types of descriptors;
    - train model;
    - calculate the most important metrics and save them into special dictionary;
    - plot true values vs predicted values 
2. Assess the results of following algorithms:
    - Linear regression (Ridge);
    - Random Forest;
    - Gradient Boosting as LGBMBoost;
    - Support Vector Machines;
    - K-Nearest Neighborrs
3. Creare dataframe, that describes model performance


```python
import pandas as pd
import numpy as np
import pickle
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import GridSearchCV, train_test_split, cross_val_score, cross_val_predict
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score, root_mean_squared_error, mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.svm import SVR
from hyperopt import tpe, Trials, hp, fmin, STATUS_OK
from functools import partial
from sklearn.neighbors import KNeighborsRegressor
```

First, let's PaDEL and RDKit datasets and EACN-target from files, that were created earlier:

```python
with open('X_padel_dropped.pickle', 'rb') as input:
    X_padel_dropped = pickle.load(input)
```

```python
with open('X_standard_dropped.pickle', 'rb') as input:
    X_standard_dropped = pickle.load(input)
```

```python
with open('EACN.pickle', 'rb') as input:
    target = pickle.load(input)
```

To facilitate automatic calculation of metrics and asiggning labels, we weill create dictionary with imported descriptors datasets

```python
descriptors_dict = {'Standard':X_standard_dropped, 'Padel':X_padel_dropped}
```

The function below performs multiple tasks:
1. First, it create matplotlib figure object and axes with two fields,
2. Then for each of two datasets  we:
    - Create a pipeline, made of StandardScaler and algorithm,
    - Split dataset into training and testing subsets;
    - Then we search for optimal parameters with functions from [hyperopt package](http://hyperopt.github.io/hyperopt/);
    - Then we extract the best model and train it with X_train and y_train;
    - Then we calculate different metrics ($R^{2}$, RMSE, MAE) and append it to the special dictionary with results, called scores, as well as some information abour descriptors sets and name of studied algorithm
    - Finally, we plot true EACN values vs predicted EACN values

```python
scores = {'Descriptors':[], 'Estimator': [], 'R2_cv':[], 'MAE_cv':[], 'RMSE_cv':[],
          'R2_test':[], 'RMSE_test':[], 'MAE_test':[], 'R2_train':[], 'RMSE_train':[], 'MAE_train':[] }
```

Special sub-function get_best_model_from_trials retrieve the model with the minumum loss from *Trials* instance

```python
def get_best_model_from_trials(trials):
    valid_trial_list = [trial for trial in trials if trial['result']['status'] == STATUS_OK]
    losses = [ float(trial['result']['loss']) for trial in valid_trial_list]
    index_having_minumum_loss = np.argmin(losses)
    best_trial_obj = valid_trial_list[index_having_minumum_loss]
    return best_trial_obj['result']
```

```python
def calculate_to_score(estimator, estimator_name, param_grid, descriptors_dict = descriptors_dict, target=target, type = 'Full'):
    fig, axes = plt.subplots(1, 2, figsize = (12,8))
    i = 0
    #for each of two datasets
    for key, value in descriptors_dict.items():
        X_train, X_test, y_train, y_test = train_test_split(value, target, random_state = 0)
        #objective function for optimization of hyperparameters
        def objective(params, X, target):
            pipe = make_pipeline(StandardScaler(), estimator)
            pipe.set_params(**params)
            y_cv_pred = cross_val_predict(pipe, X, target, cv=5)
            cv_rmse = np.sqrt(root_mean_squared_error(target, y_cv_pred))
            cv_r2 = r2_score(target, y_cv_pred)
            
            score = cross_val_score(pipe, X, target, cv = 10, scoring = 'neg_mean_absolute_error')
            return {'loss':-score.mean(), 'status':STATUS_OK, 'Trained_model':pipe, 'rmse':cv_rmse, 'r2':cv_r2}
        trials = Trials()
        #get best hyperparameters
        best = fmin(partial(objective, X = X_train, target = y_train), space = param_grid, trials = trials, algo = tpe.suggest, max_evals=  150)
        print(best)
        #get best model
        best_estimator  = get_best_model_from_trials(trials)['Trained_model']
      
        #calculate metrics
        cv_r2 = get_best_model_from_trials(trials)['r2']
        cv_rmse = get_best_model_from_trials(trials)['rmse']
        cv_mae = get_best_model_from_trials(trials)['loss']
        
        scores['R2_cv'].append(cv_r2)
        scores['RMSE_cv'].append(cv_rmse)
        scores['MAE_cv'].append(cv_mae)
        
        best_estimator.fit(X_train, y_train)
        predictions = best_estimator.predict(X_test)
        predictions_train = best_estimator.predict(X_train)
        scores['Descriptors'].append(key)
        scores['Estimator'].append(estimator_name)
        scores['R2_test'].append(r2_score(y_test, predictions))
        scores['R2_train'].append(r2_score(y_train, predictions_train))
        
        scores['RMSE_test'].append(root_mean_squared_error(y_test, predictions))
        scores['RMSE_train'].append(root_mean_squared_error(y_train, predictions_train))

        
        scores['MAE_test'].append(mean_absolute_error(y_test, predictions))
        scores['MAE_train'].append(mean_absolute_error(y_train, predictions_train))
        #plot results
        axes[i].plot(y_test, predictions, 'o')
        axes[i].plot([0,40], [0,40])
        axes[i].set_title('True vs predicted values for {} descriptors'.format(key))
        axes[i].set_xlabel('True test values')
        axes[i].set_ylabel('Predicted test values')
        i += 1
    plt.tight_layout()
```

Let's initialize scores dictionary and test function with Logistic Regression

```python
calculate_to_score(Ridge(max_iter = 10000), estimator_name='Linear Regression', 
                   param_grid = {'ridge__alpha':hp.loguniform(label = 'alpha', low = -4*np.log(10), high  = 2*np.log(10)),
                                'ridge__solver':hp.choice(label = 'solver', options = ['auto', 'svd', 'cholesky', 'lsqr', 'sparse_cg','sag', 'saga'])})
```
